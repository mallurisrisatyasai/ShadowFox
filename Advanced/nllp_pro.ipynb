```python
import os
from dotenv import load_dotenv
import google.generativeai as genai
import time
```

```python
print("GOOGLE GEMINI API - LATEST MODELS")
# Load API key
load_dotenv()
api_key = os.getenv('GOOGLE_API_KEY')

if not api_key:
    print("\n API KEY NOT FOUND!")
    print(" SOLUTION:")
    print("1. Create .env file")
    print("2. Add: GOOGLE_API_KEY=AQ.Ab8RN6K...your_key...")
    print("3. Save and restart")
    exit()

# Configure API
genai.configure(api_key=api_key)

print("\n API KEY LOADED!")
print(" Using latest Gemini models!")
```

```python
print("\n📋 Checking available models...\n")

try:
    models = genai.list_models()
    print("Available models:")
    for model in models:
        if 'generateContent' in model.supported_generation_methods:
            print(f"  ✓ {model.name}")
except Exception as e:
    print(f"Note: {e}")

print("\nUsing: gemini-2.0-flash (or gemini-1.5-flash as backup)")
```

```python
def generate_text(prompt, model_name="gemini-3.1-flash-lite", max_tokens=500):
    """Generate text using Google Gemini"""
    try:
        model = genai.GenerativeModel(model_name)
        response = model.generate_content(prompt)
        return response.text
    except Exception as e:
        print(f"Error with {model_name}: {e}")
        print("Trying gemini-1.5-flash...")
        try:
            model = genai.GenerativeModel("gemini-1.5-flash")
            response = model.generate_content(prompt)
            return response.text
        except Exception as e2:
            return f"Error: {e2}"
```

```python
print("TEST 1: SIMPLE QUESTION")

prompt = "What is artificial intelligence?"
print(f"\n Q: {prompt}\n")

start = time.time()
response = generate_text(prompt)
elapsed = time.time() - start

print(f" A: {response[:300]}...")
print(f"  Time: {elapsed:.2f}s")

```

```python
print("TEST 2: TEXT SUMMARIZATION")

text = """
Machine learning is a subset of artificial intelligence that focuses on enabling 
computers to learn from data without explicit programming. Deep learning uses neural 
networks with multiple layers to process information. Natural language processing 
allows computers to understand and generate human language. These technologies are 
transforming industries like healthcare, finance, and transportation.
"""

summarize_prompt = f"Summarize this in 2 sentences:\n\n{text}"
print(f"\n Summarizing text...\n")

start = time.time()
summary = generate_text(summarize_prompt)
elapsed = time.time() - start

print(f" Summary: {summary}")
print(f"  Time: {elapsed:.2f}s")
```

```python
print("TEST 3: CODE GENERATION")

code_prompt = "Write Python code to calculate factorial of a number"
print(f"\n Generating code...\n")

start = time.time()
code = generate_text(code_prompt)
elapsed = time.time() - start

print(f" Code:\n{code}")
print(f"  Time: {elapsed:.2f}s")
```

```python
print("TEST 4: SENTIMENT ANALYSIS")

texts = [
    "This product is absolutely amazing!",
    "Terrible quality, very disappointed.",
    "It's okay, nothing special."
]

print(f"\nAnalyzing {len(texts)} texts...\n")

for text in texts:
    sentiment_prompt = f"Analyze sentiment (Positive/Negative/Neutral):\n{text}"
    result = generate_text(sentiment_prompt, max_tokens=50)
    print(f" Text: {text}")
    print(f" Result: {result[:100]}\n")
```

```python
print("TEST 5: CREATIVE WRITING")

creative_prompt = "Write a short funny story about a robot learning to cook (100 words)"
print(f"\n  Generating story...\n")

start = time.time()
story = generate_text(creative_prompt)
elapsed = time.time() - start

print(f" Story:\n{story}")
print(f"  Time: {elapsed:.2f}s")
```

```python
print("TEST 6: TRANSLATION")

translate_prompt = "Translate to Spanish: 'Hello, how are you today?'"
print(f"\n Translating...\n")

start = time.time()
translation = generate_text(translate_prompt)
elapsed = time.time() - start

print(f" Translation: {translation}")
print(f"  Time: {elapsed:.2f}s")
```

```python
print("TEST 7: QUESTION ANSWERING")


context = "Python was created by Guido van Rossum in 1989. It emphasizes code readability."
qa_prompt = f"Context: {context}\n\nQuestion: When was Python created?\nAnswer:"

print(f"\nAnswering question from context...\n")

start = time.time()
answer = generate_text(qa_prompt)
elapsed = time.time() - start

print(f" Answer: {answer}")
print(f"  Time: {elapsed:.2f}s")
```

```python
print("TEST 8: TEXT CLASSIFICATION")

classify_prompt = "Classify as Technology/Sports/Business: 'New AI model achieves breakthrough'"
print(f"\nClassifying text...\n")

start = time.time()
category = generate_text(classify_prompt, max_tokens=50)
elapsed = time.time() - start

print(f" Category: {category}")
print(f"  Time: {elapsed:.2f}s")
```
